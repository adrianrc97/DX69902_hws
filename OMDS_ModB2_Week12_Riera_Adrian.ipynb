# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

The first option makes it easier to see the changes during each week separately. However, it is harder to compare the number of visitors between the two weeks.

The second option puts both weeks on the same graph, making it easier to compare the number of visitors on each day. The disadvantage is that the lines can become confusing if we have too much data.

I would choose the second option because it makes comparing the two weeks easier. I would also add a legend and label the days of the week.

```python
import numpy as np
import matplotlib.pyplot as plt

years = [2008, 2009, 2010, 2011, 2012, 2013, 2014]

prices = {
    "A": [390, 420, 430, 390, 300, 270, 260],
    "B": [360, 400, 410, 370, 290, 250, 240],
    "C": [np.nan, np.nan, 100, 180, 190, 230, 180],
    "D": [np.nan, np.nan, np.nan, 140, 260, 220, 210],
    "E": [np.nan, np.nan, np.nan, np.nan, np.nan, 110, 200]
}

plt.figure(figsize=(10, 5))
plt.axhspan(150, 200, color="lightsteelblue", alpha=0.4)

for product, values in prices.items():
    plt.plot(years, values, color="gray", linewidth=2)
    for i in range(len(values)):
        if not np.isnan(values[i]):
            plt.text(years[i] - 0.35, values[i], product, color="dimgray")
            break

plt.scatter(2014, 223, color="navy", s=35)
plt.text(2014.1, 223, "AVG $223", color="navy")
plt.text(2008, 175, "Recommended range", color="navy")

plt.title("Recommended product price: $150-$200", fontsize=14, color="navy")
plt.xlabel("Year")
plt.ylabel("Average retail price")
plt.xticks(years)
plt.yticks([0, 100, 200, 300, 400, 500], ["$0", "$100", "$200", "$300", "$400", "$500"])
plt.ylim(0, 500)
plt.xlim(2007, 2015)
plt.grid(axis="y", alpha=0.2)
plt.show()
```
